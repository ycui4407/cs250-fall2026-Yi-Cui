### Question 4 (Medium — 10 points)

Using nested loops and the flag-variable pattern from lecture — including the **square-root optimization** (only testing possible divisors up to `int(candidate ** 0.5) + 1`) — build a list `primes_in_range` containing every prime number between `50` and `80`, inclusive.

**Important note:** a number is prime only if it is greater than 1 and has no divisors other than 1 and itself. Reset your flag variable for **every new candidate** inside the outer loop — reusing a flag from the previous candidate is a common bug.

```python
# Write your code below
primes_in_range = []

for candidate in range(50, 81):
    is_prime = True

    for divisor in range(2, int(candidate ** 0.5) + 1):
        if candidate % divisor == 0:
            is_prime = False
            break

    if is_prime:
        primes_in_range.append(candidate)
```

```python
# ✅ Check Question 4
def _check_q4():
    if primes_in_range != [53, 59, 61, 67, 71, 73, 79]:
        print(f"❌ `primes_in_range` is {primes_in_range!r}, expected [53, 59, 61, 67, 71, 73, 79].\n"
              f"   Hint: for candidate in range(50, 81):\n"
              f"             is_prime = True\n"
              f"             limit = int(candidate ** 0.5) + 1\n"
              f"             for i in range(2, limit):\n"
              f"                 if candidate % i == 0:\n"
              f"                     is_prime = False\n"
              f"                     break\n"
              f"             if is_prime:\n"
              f"                 primes_in_range.append(candidate)")
        return 0
    print("✅ Correct! +10 points")
    return 10

SCORE["q4"] = _check_q4()
```

### Question 7 (Medium — 10 points)

Given the integer `number` below, check whether it's a palindrome **without ever converting it to a string** — use the arithmetic digit-reversal technique from lecture (repeatedly pull off the last digit with `% 10` and rebuild the reversed number). Write code that:

2. Stores in `is_palindromic_number` a Boolean: `True` if `number` equals `reversed_number`, `False` otherwise.

**Important note:** this number is **not** a palindrome — make sure your code correctly produces `False` rather than assuming every test case will match. Use a separate working copy of the number inside your loop so `number` itself stays unchanged.

```python
number = 6284

# Write your code below
reversed_number = 0
is_palindromic_number = None

original = number
n = number
while n > 0:
    reversed_number = reversed_number * 10 + n % 10
    n //= 10
is_palindromic_number = original == reversed_number
print(is_palindromic_number)

```

```python
# ✅ Check Question 7
def _check_q7():
    ok = True
    if reversed_number != 4826:
        print(f"❌ `reversed_number` is {reversed_number!r}, expected 4826.\n"
              f"   Hint: use a working copy, e.g. n = number, then while n > 0:\n"
              f"             reversed_number = reversed_number * 10 + n % 10\n"
              f"             n //= 10")
        ok = False
    if is_palindromic_number is not False:
        print(f"❌ `is_palindromic_number` is {is_palindromic_number!r}, expected False (6284 reversed is 4826, not equal).\n"
              f"   Hint: is_palindromic_number = (number == reversed_number)")
        ok = False
    if number != 6284:
        print(f"❌ `number` changed to {number!r} — it should stay 6284. Use a separate working variable "
              f"(like `n = number`) inside your loop instead of modifying `number` directly.")
        ok = False
    if ok:
        print("✅ Correct! +10 points")
        return 10
    return 0

SCORE["q7"] = _check_q7()
```

### Question 8 (Medium — 10 points)

Given the phrase `phrase = "Never Odd or Even"` below, write code that:

1. Builds `cleaned` by looping through `phrase.lower()` one character at a time and keeping only the characters where `.isalnum()` is `True` (this removes spaces).
2. Uses the **two-pointer technique** (not slicing) — a `left` index starting at 0, a `right` index starting at the last position, and a `while left < right:` loop comparing characters and moving inward — to determine whether `cleaned` is a palindrome. Store the result in `is_palindrome_phrase`.

**Important note:** use `break` the moment you find a mismatch, but don't forget that if the loop finishes without ever breaking, the phrase **is** a palindrome. Initialize your flag to `True` before the loop, exactly like the flag pattern used for primality testing.

```python
phrase = "Never Odd or Even"

# Write your code below
cleaned = ""

for ch in phrase.lower():
    if ch.isalnum():
       cleaned += ch

is_palindrome_phrase = True
left = 0
right = len(cleaned)-1

while left < right:
    if cleaned[left] != cleaned[right]:
        is_palindrome_phrase = False
        break
    left += 1
    right -= 1

print(is_palindrome_phrase)
```

```python
# ✅ Check Question 8
def _check_q8():
    ok = True
    if cleaned != "neveroddoreven":
        print(f"❌ `cleaned` is {cleaned!r}, expected 'neveroddoreven'.\n"
              f"   Hint: for ch in phrase.lower():\n"
              f"             if ch.isalnum():\n"
              f"                 cleaned += ch")
        ok = False
    if is_palindrome_phrase is not True:
        print(f"❌ `is_palindrome_phrase` is {is_palindrome_phrase!r}, expected True.\n"
              f"   Hint: right = len(cleaned) - 1, then while left < right: compare cleaned[left] and cleaned[right], "
              f"move left forward and right backward each time they match.")
        ok = False
    if ok:
        print("✅ Correct! +10 points")
        return 10
    return 0

SCORE["q8"] = _check_q8()
```

2. Stores in `stock_status` a list the **same length** as `inventory_counts`, where each count becomes the string `"low"` if it's less than 20, otherwise `"ok"` (a conditional-expression comprehension: `if...else` placed **before** the `for`).

**Important note:** a trailing `if` filters items out of the result, so the output can be shorter than the input. An `if...else` placed before the `for` instead chooses between two output values for every item, so the output is always the same length as the input — these are two different tools, not interchangeable.

```python
inventory_counts = [12, 45, 7, 90, 3, 61, 28]

# Write your code below
low_stock = [count for count in inventory_counts if count < 20]
stock_status = ["low" if count < 20 else "ok" for count in inventory_counts]
```

```python
# ✅ Check Question 9
def _check_q9():
    ok = True
    if low_stock != [12, 7, 3]:
        print(f"❌ `low_stock` is {low_stock!r}, expected [12, 7, 3].\n"
              f"   Hint: low_stock = [count for count in inventory_counts if count < 20]")
        ok = False
    if stock_status != ["low", "ok", "low", "ok", "low", "ok", "ok"]:
        print(f"❌ `stock_status` is {stock_status!r}, expected ['low', 'ok', 'low', 'ok', 'low', 'ok', 'ok'].\n"
              f'   Hint: stock_status = ["low" if count < 20 else "ok" for count in inventory_counts]')
        ok = False
    if ok:
        print("✅ Correct! +10 points")
        return 10
    return 0

SCORE["q9"] = _check_q9()
```

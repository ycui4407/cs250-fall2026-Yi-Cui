A restaurant ordering program that helps customers to order food and calculate the total price.

```python
def display_menu():
    """Display the food menu with prices and return the menu dictionary."""
    menu = {
        "Double-Double": 4.45,
        "Hamburger": 2.75,
        "Cheeseburger": 3.10,
        "French Fries": 2.05,
        "Drink": 2.05
    }

    print("Welcome to In-N-Out!\n")
    print("Menu:")

    for item, price in menu.items():
        print(f"{item}: ${price:.2f}")
    print("\n")
    return menu



def get_order(menu):
    """Ask the customer to choose food items and quantity."""
    order = {}
    add_item = "yes"
    lowercase_mapping = {item.lower(): item for item in menu}

    while add_item == "yes":
        print("Please choose an item:")
        choice = input("Enter food name: ").strip().lower()

        if choice in lowercase_mapping:
            choice = lowercase_mapping[choice]
        else:
            print("not on menu.")
            continue
        try:
            quantity = int(input("Enter quantity: "))
            if quantity <= 0:
                print("Invalid order. Quantity must be a positive integer.")
                continue
        except ValueError:
            print("Invalid order. Quantity must be an integer.")
            continue

        order[choice] = order.get(choice, 0) + quantity
        add_item = input("Add another item? (yes/no): ").lower()

    return order


def calculate_total(subtotal, tax_rate=0.09):
    """Calculate the total price including tax."""
    tax = subtotal * tax_rate
    total = subtotal + tax
    return total, tax

def process_order(order, menu):
    """Process the order and calculate the subtotal."""
    print("\nOrder Summary\n")
    subtotal = 0
    for item, quantity in order.items():
        price = menu[item]
        item_subtotal = price * quantity
        print(f"{item}: {quantity} x ${price:.2f} = ${item_subtotal:.2f}")
        subtotal += item_subtotal
    total, tax = calculate_total(subtotal, tax_rate=0.09)
    show_order_summary(subtotal, total, tax)

def show_order_summary(subtotal, total, tax):
    """Display the items ordered and the total price."""
    print(f"Subtotal: ${subtotal:.2f}")
    print(f"Tax: ${tax:.2f}")
    print(f"Total_after_tax: ${total:.2f}")
    print("\nThank you for your order!\nWe appreciate your business.")
```

```python
def main():
    menu = display_menu()
    order = get_order(menu)
    process_order(order, menu)
main()
```

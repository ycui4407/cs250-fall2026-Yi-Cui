In-N-Out Ordering System

A simple restaurant ordering program that allows customers to choose food items, enter quantities, customize their orders, and calculate the total price.

```python
# Imports
# No additional modules are needed.
```

```python
def display_menu():
    """Display the food menu with prices and return the menu dictionary."""
    menu = {
        "Double-Double": 4.45,
        "Hamburger": 2.75,
        "Cheeseburger": 3.10,
        "French Fries": 2.05,
        "Drink": 2.05
    }

    print("Welcome to In-N-Out!")
    print("Menu:")

    for item, price in menu.items():
        print(f"{item}: ${price:.2f}")
    print("\n")
    return menu



def get_order(menu):
    """Ask the customer to choose food items and quantity."""
    order = {}
    add_item = "yes"

    while add_item == "yes":
        print("Please choose an item:")


        choice = input("Enter food name: ")

        if choice in menu:
            item = choice
        else:
            print("not on menu.")
            continue

        quantity = int(input("Enter quantity: "))

        order[item] = order.get(item,0) + quantity

        add_item = input("Add another item? (yes/no): ").lower()

    return order



def calculate_subtotal(price, quantity):
    """Calculate the subtotal for one food item."""
    item_subtotal = price * quantity
    return item_subtotal



def calculate_total(subtotal, tax_rate=0.09):
    """Calculate the total price including tax."""
    tax = subtotal * tax_rate
    total = subtotal + tax
    return total, tax


def show_order_summary(order, subtotal, total, menu, tax):
    """Display the items ordered and the total price."""
    print("\nOrder Summary")

    for item, quantity in order.items():
        price = menu[item]
        item_subtotal = price * quantity
        print(f"{item} x {quantity}: ${item_subtotal:.2f}")

    print(f"Subtotal: ${subtotal:.2f}")
    print(f"Tax: ${tax:.2f}")
    print(f"Total_after_tax: ${total:.2f}")
    print("\nThank you for your order!\nWe appreciate your business.")
```

```python
def main():
    menu = display_menu()
    order = get_order(menu)

    subtotal = 0
    for item, quantity in order.items():
        price = menu[item]
        item_subtotal = calculate_subtotal(price, quantity)
        subtotal += item_subtotal

    total, tax = calculate_total(subtotal, tax_rate=0.09)

    show_order_summary(order, subtotal, total, menu, tax)

main()
```

### Question 6

Two work crews each carry a set of tools, shown below. Write code that stores:

1. `shared_tools` — tools **both** crews carry (intersection).
2. `all_tools_used` — every tool carried by **either** crew (union).
3. `morning_only` — tools the morning crew carries that the night crew does **not** (difference).
4. `exclusive_tools` — tools carried by **exactly one** crew, combining two differences: everything in the morning crew but not the night crew, **plus** everything in the night crew but not the morning crew.

**Important note:** set difference is not symmetric (`a - b` is not the same as `b - a`), so `exclusive_tools` needs both directions combined with `|`. This chained expression is exactly what programmers mean by a "symmetric difference" — the parts that do not overlap.

```python
morning_crew = {"drill", "ladder", "level", "hammer"}
night_crew = {"level", "hammer", "saw"}

# Write your code below
shared_tools = morning_crew & night_crew
all_tools_used = morning_crew | night_crew
morning_only = morning_crew - night_crew
exclusive_tools = (morning_crew - night_crew)|(night_crew - morning_crew)
```

```python
# ✅ Check Question 6
def _check_q6():
    ok = True
    if shared_tools != {"level", "hammer"}:
        print(f"❌ `shared_tools` is {shared_tools!r}, expected {{'level', 'hammer'}}.\n   Hint: shared_tools = morning_crew & night_crew")
        ok = False
    if all_tools_used != {"drill", "ladder", "level", "hammer", "saw"}:
        print(f"❌ `all_tools_used` is {all_tools_used!r}.\n   Hint: all_tools_used = morning_crew | night_crew")
        ok = False
    if morning_only != {"drill", "ladder"}:
        print(f"❌ `morning_only` is {morning_only!r}, expected {{'drill', 'ladder'}}.\n   Hint: morning_only = morning_crew - night_crew")
        ok = False
    if exclusive_tools != {"drill", "ladder", "saw"}:
        print(f"❌ `exclusive_tools` is {exclusive_tools!r}, expected {{'drill', 'ladder', 'saw'}}.\n"
              f"   Hint: exclusive_tools = (morning_crew - night_crew) | (night_crew - morning_crew)")
        ok = False
    if ok:
        print("✅ Correct! +12 points")
        return 12
    return 0

SCORE["q6"] = _check_q6()
```

### Question 8

An arcade counts tickets awarded per game round, stored in the tuple `ticket_batches` below. Write a **while loop** (not a for loop) that uses an index variable, starting at `0`, to compute both of the following in a single pass:

1. `total_tickets` — the sum of every batch.
2. `big_batches` — the number of batches with `50` or more tickets.

**Important note:** a `while` loop needs all four parts working together — an initialized index, a condition comparing it to `len(ticket_batches)`, a body that updates both an accumulator and a counter, and an increment to the index so the loop eventually stops.

```python
ticket_batches = (25, 60, 15, 80, 40, 5)

# Write your code below
total_tickets = 0
big_batches = 0
index = 0

while index < len(ticket_batches):
    total_tickets += ticket_batches[index]

    if ticket_batches[index] >= 50:
        big_batches += 1

    index += 1
```

```python
# ✅ Check Question 8
def _check_q8():
    ok = True
    if total_tickets != 225:
        print(f"❌ `total_tickets` is {total_tickets!r}, expected 225.\n"
              f"   Hint: total_tickets += ticket_batches[index] inside the loop.")
        ok = False
    if big_batches != 2:
        print(f"❌ `big_batches` is {big_batches!r}, expected 2 (60 and 80 both qualify).\n"
              f"   Hint: while index < len(ticket_batches):\n"
              f"             if ticket_batches[index] >= 50:\n"
              f"                 big_batches += 1\n"
              f"             total_tickets += ticket_batches[index]\n"
              f"             index += 1")
        ok = False
    if ok:
        print("✅ Correct! +12 points")
        return 12
    return 0

SCORE["q8"] = _check_q8()
```

```python
# ✅ Check Question 9 — edge-case test battery
def _check_q9():
    pts = 0
    checks_passed = 0
    checks_total = 9

    expected_records = [
        ("Burrito", 3, 8.50, 25.50),
        ("Nachos", 1, 65.00, 65.00),
        ("Burrito", 2, 8.50, 17.00),
        ("Chips", 0, 3.00, 0.0),
        ("Catering Tray", 2, 50.0, 100.0),
    ]
    expected_unique_items = {"Burrito", "Nachos", "Chips", "Catering Tray"}
    expected_totals_by_item = {"Burrito": 42.5, "Nachos": 65.0, "Chips": 0.0, "Catering Tray": 100.0}
    expected_zero_qty_items = ["Chips"]
    expected_items_list = ["Burrito", "Nachos", "Burrito", "Chips", "Catering Tray"]
    expected_size_labels = ["small", "medium", "small", "small", "large"]
    expected_sorted_totals = [100.0, 65.0, 25.5, 17.0, 0.0]
    expected_summary = f"Processed {len(order_log)} orders across 4 items; 1 large order(s)."

    # Test 1: records — tuple structure and values (including the repeated-item rows kept separate)
    try:
        ok = records == expected_records
    except Exception:
        ok = False
    if ok:
        checks_passed += 1
        print("✅ Test 1/9 — `records` (tuples, correct structure): PASS")
    else:
        print(f"❌ Test 1/9 — `records` is {records!r}\n"
              f"   Expected {expected_records!r}\n"
              f"   Hint: for each order line, split on \",\", convert quantity to int and price to float, "
              f"then append (item, quantity, price, line_total) as a tuple.")

    # Test 2: unique_items — set dedupes the repeated "Burrito"
    ok = unique_items == expected_unique_items
    if ok:
        checks_passed += 1
        print("✅ Test 2/9 — `unique_items` (set dedupes repeats): PASS")
    else:
        print(f"❌ Test 2/9 — `unique_items` is {unique_items!r}, expected {expected_unique_items!r}\n"
              f"   Hint: add each item name to a set as you loop — a set automatically drops the duplicate \"Burrito\".")

    # Test 3: totals_by_item — the repeated item accumulates rather than overwrites
    ok = (isinstance(totals_by_item, dict)
          and set(totals_by_item.keys()) == set(expected_totals_by_item.keys())
          and all(abs(totals_by_item.get(k, 0) - v) < 1e-9 for k, v in expected_totals_by_item.items()))
    if ok:
        checks_passed += 1
        print("✅ Test 3/9 — `totals_by_item` (repeated item accumulates): PASS")
    else:
        print(f"❌ Test 3/9 — `totals_by_item` is {totals_by_item!r}\n"
              f"   Expected {expected_totals_by_item!r}\n"
              f"   Hint: this is the classic edge case — \"Burrito\" was ordered on two separate lines (25.50 and 17.00), "
              f"so its total should be 42.50, not just the last order's value. Use "
              f"totals_by_item[item] = totals_by_item.get(item, 0) + line_total.")

    # Test 4: zero_qty_items — the quantity=0 line for Chips is flagged, not skipped
    ok = zero_qty_items == expected_zero_qty_items
    if ok:
        checks_passed += 1
        print("✅ Test 4/9 — `zero_qty_items` (zero-quantity edge case): PASS")
    else:
        print(f"❌ Test 4/9 — `zero_qty_items` is {zero_qty_items!r}, expected {expected_zero_qty_items!r}\n"
              f"   Hint: check `if quantity == 0:` inside the loop and append the item name — the Chips order must "
              f"still be processed like any other, just flagged.")

    # Test 5: items_list + has_nachos — searching a plain list (not the set)
    ok = items_list == expected_items_list and has_nachos is True
    if ok:
        checks_passed += 1
        print("✅ Test 5/9 — `items_list` / `has_nachos` (list search): PASS")
    else:
        print(f"❌ Test 5/9 — `items_list` is {items_list!r}, `has_nachos` is {has_nachos!r}\n"
              f"   Expected items_list={expected_items_list!r}, has_nachos=True\n"
              f"   Hint: items_list.append(item) inside the loop, then has_nachos = \"Nachos\" in items_list.")

    # Test 6: size_labels — if/elif/else classification, including the exact-100 boundary
    ok = size_labels == expected_size_labels
    if ok:
        checks_passed += 1
        print("✅ Test 6/9 — `size_labels` (if/elif/else + boundary case): PASS")
    else:
        print(f"❌ Test 6/9 — `size_labels` is {size_labels!r}\n"
              f"   Expected {expected_size_labels!r}\n"
              f"   Hint: the Catering Tray total is exactly 100.0 — with `if line_total < 50: ... elif line_total < 100: ... "
              f"else: ...`, exactly 100 falls into the else branch (\"large\"), not \"medium\".")

    # Test 7: large_order_count — boundary uses >=, not >
    ok = large_order_count == 1
    if ok:
        checks_passed += 1
        print("✅ Test 7/9 — `large_order_count` (boundary uses >=): PASS")
    else:
        print(f"❌ Test 7/9 — `large_order_count` is {large_order_count!r}, expected 1\n"
              f"   Hint: only the Catering Tray order (line_total == 100.0) qualifies with `line_total >= 100`.")

    # Test 8: sorted_totals — sorting a list of numbers, descending
    ok = sorted_totals == expected_sorted_totals
    if ok:
        checks_passed += 1
        print("✅ Test 8/9 — `sorted_totals` (sort descending): PASS")
    else:
        print(f"❌ Test 8/9 — `sorted_totals` is {sorted_totals!r}\n"
              f"   Expected {expected_sorted_totals!r}\n"
              f"   Hint: sorted_totals = sorted(all_the_line_totals, reverse=True)")

    # Test 9: summary_line — f-string pulling together earlier results
    ok = summary_line == expected_summary
    if ok:
        checks_passed += 1
        print("✅ Test 9/9 — `summary_line` (f-string formatting): PASS")
    else:
        print(f"❌ Test 9/9 — `summary_line` is {summary_line!r}\n"
              f"   Expected {expected_summary!r}\n"
              f"   Hint: build it from variables you already have, e.g.\n"
              f'   summary_line = f"Processed {{len(order_log)}} orders across {{len(unique_items)}} items; '
              f'{{large_order_count}} large order(s)."')

    print(f"\n{checks_passed}/{checks_total} edge-case tests passed.")
    pts = round(20 * checks_passed / checks_total)
    if checks_passed == checks_total:
        print(f"✅ All edge cases handled! +{pts} points")
    elif checks_passed > 0:
        print(f"⚠️ Partial credit: +{pts} points")
    else:
        print("❌ No tests passed yet — 0 points")
    return pts

SCORE["q9"] = _check_q9()
```
